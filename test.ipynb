```python
import yfinance as yf

def get_trading_api(stock):
    df = yf.download(stock, period="5y", interval="1d")
    df.reset_index(inplace=True)
    df['Stock'] = stock
    return df
```

```python
get_trading_api("TCS.NS")
```
